### **DATA CLEANING OF THE ITEM FAT CONTENT COLUMN:** 

```python
df["Item Fat Content"]=df["Item Fat Content"].replace("low fat","Low Fat")
```

```python
df["Item Fat Content"]=df["Item Fat Content"].replace({"LF":"Low Fat","reg":"Regular"})
```

```python
df["Item Fat Content"].unique()
```

## **BUSINESS REQUIREMENTS**

```python
Total_Sales=df["Sales"].sum()
print("Total Sales is ",Total_Sales)
Average_Sales=df["Sales"].mean()
print("Average Sales is :",Average_Sales)
```

```python
Average_Rating=df["Rating"].mean()
print("Average Rating is :",Average_Rating)
```

```python
No_of_Items_Sold=df["Rating"].count()
print("No_of_Items_Sold is :",No_of_Items_Sold)
```

```python
print(f"Average Sales is: ${Average_Sales:,.2f}")
```

### **CHARTS** 

```python
fat_sales=df.groupby("Item Fat Content")["Sales"]
```

```python
for x in fat_sales:
    print(x)
```

```python
sales_by_fat=fat_sales.sum()
```

```python
sales_by_fat
```

```python
plt.pie(sales_by_fat,labels=sales_by_fat.index,autopct="%.1f%%",startangle=135)
plt.title("Sales by fat content of the items")
plt.axis("equal")
plt.show()
```

## **total sales by item type**

```python
Sales_ItemType=df.groupby("Item Type")['Sales'].sum().sort_values(ascending=False)
```

```python
plt.pie(Sales_ItemType,labels=Sales_ItemType.index,autopct="%.1f%%",startangle=45)
plt.title("Sales per Item Type")
plt.axis("equal")
plt.show()
```

```python
Sales_byItem=df.groupby("Item Type")["Sales"].sum().sort_values(ascending=False)
```

```python
plt.bar(Sales_byItem.index,Sales_byItem.values)
plt.title("Sales on basis of Item types")
plt.xlabel("Item Type")
plt.ylabel("Total Sales")
plt.xticks(rotation=-90)
#plt.tight_layout
plt.show()
```

```python
grouped=df.groupby(["Outlet Location Type","Item Fat Content"])["Sales"].sum().unstack()
```

```python
grouped.plot(kind="bar")
plt.title("Sales corresponding to outlet location and Item Fat Content")
plt.xticks(rotation=90)
plt.ylabel("Sales")
plt.xlabel("Outlet Location Type")

plt.show()
```

```python
df.columns
```

```python
Year_Sales=df.groupby("Outlet Establishment Year")["Sales"].sum().sort_index()
```

```python
plt.plot(Year_Sales.index,Year_Sales.values)
plt.xlabel("Outlets Establishment Year")
plt.ylabel("Sales")
plt.title("SALES CORRESPONDING TO THE OUTLETS ESTABLISHMENT YEAR")
plt.xticks(rotation=0)
plt.show()
```

```python
Sales_OutletLocation=df.groupby("Outlet Location Type")["Sales"].sum().sort_values()
```

```python
sns.barplot(x=Sales_OutletLocation.index,y=Sales_OutletLocation.values)
plt.xlabel("Outlet Location")
plt.ylabel("Sales per outlet location")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
